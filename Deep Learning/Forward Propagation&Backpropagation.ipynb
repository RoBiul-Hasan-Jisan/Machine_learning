# Forward Propagation & Backpropagation 

## 1. The Big Picture
Training a neural network is a cycle of two main events:
1.  **Forward Pass (Prediction):** The data flows from input to output. The network makes a guess.
2.  **Backward Pass (Learning):** We measure how wrong the guess was (Loss) and propagate that error backward to update the weights using the **Chain Rule**.

---

## 2. Forward Propagation
This is simply the calculation of the output layer by layer.

**The Setup:**
Imagine a simple **2-Layer Network** (1 Hidden Layer, 1 Output Layer):
* **Input:** $X$
* **Weights:** $W_1$ (Input $\to$ Hidden), $W_2$ (Hidden $\to$ Output)
* **Biases:** $b_1, b_2$
* **Activation Function:** $\sigma$ (e.g., Sigmoid or ReLU)

**The Steps:**
1.  **Hidden Layer Calculation:**
    $$Z_1 = W_1 \cdot X + b_1$$
    $$A_1 = \sigma(Z_1)$$
2.  **Output Layer Calculation:**

    $$Z_2 = W_2 \cdot A_1 + b_2$$
    $$A_2 = \sigma(Z_2)$$
     (This is our prediction $\hat{y}$)$$
    

3.  **Loss Calculation (Error):**
    We compare prediction $A_2$ vs actual label $y$. Let's use **Mean Squared Error (MSE)** for simplicity:
    $$L = \frac{1}{2}(y - A_2)^2$$



---

## 3. Backpropagation (The Chain Rule)
This is where the magic happens. We want to know: **"How much did a specific weight ($W$) contribute to the total Error ($L$)?"**

Mathematically, we need the gradient: $\frac{\partial L}{\partial W}$.

Since $W$ is buried deep inside the equations, we use the **Chain Rule**:
$$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}$$

### The Derivation (Demystifying the Math)
Let's calculate the gradients for $W_2$ (Output Layer) and $W_1$ (Hidden Layer).



#### A. Gradients for Output Layer ($W_2$)
We want $\frac{\partial L}{\partial W_2}$. We break it down into a chain:
$$\frac{\partial L}{\partial W_2} = \frac{\partial L}{\partial A_2} \cdot \frac{\partial A_2}{\partial Z_2} \cdot \frac{\partial Z_2}{\partial W_2}$$

1.  **Change in Loss w.r.t Prediction:**
    $$\frac{\partial L}{\partial A_2} = -(y - A_2)$$
2.  **Change in Prediction w.r.t Z (Activation Derivative):**
    $$\frac{\partial A_2}{\partial Z_2} = \sigma'(Z_2)$$
3.  **Change in Z w.r.t Weight:**
    Since $Z_2 = W_2 A_1 + b_2$:
    $$\frac{\partial Z_2}{\partial W_2} = A_1$$

**Combined:**
$$\frac{\partial L}{\partial W_2} = \underbrace{-(y - A_2) \cdot \sigma'(Z_2)}_{\delta_2 \text{ (Error at Output)}} \cdot A_1$$

---

#### B. Gradients for Hidden Layer ($W_1$)
This is deeper. The error must pass *through* the output layer to get here.
$$\frac{\partial L}{\partial W_1} = \frac{\partial L}{\partial A_2} \cdot \frac{\partial A_2}{\partial Z_2} \cdot \frac{\partial Z_2}{\partial A_1} \cdot \frac{\partial A_1}{\partial Z_1} \cdot \frac{\partial Z_1}{\partial W_1}$$

Notice the first two terms are the same as before ($\delta_2$). Let's extend the chain:

1.  **Backpropagate through $W_2$:**
    Since $Z_2 = W_2 A_1 + b_2$, the change in $Z_2$ with respect to $A_1$ is $W_2$.
    $$\frac{\partial Z_2}{\partial A_1} = W_2$$
2.  **Derivative of Hidden Activation:**
    $$\frac{\partial A_1}{\partial Z_1} = \sigma'(Z_1)$$
3.  **Input to Hidden Layer:**
    Since $Z_1 = W_1 X + b_1$:
    $$\frac{\partial Z_1}{\partial W_1} = X$$

**Combined:**
$$\frac{\partial L}{\partial W_1} = \underbrace{\delta_2 \cdot W_2 \cdot \sigma'(Z_1)}_{\delta_1 \text{ (Error at Hidden)}} \cdot X$$

---

## 4. The Update Rule (Gradient Descent)
Once we have the gradients ($\frac{\partial L}{\partial W}$), we update the weights to minimize error. We move in the opposite direction of the gradient.



$$W_{new} = W_{old} - \alpha \cdot \frac{\partial L}{\partial W}$$

*Where $\alpha$ is the **Learning Rate** (step size).*

## Summary Checklist
| Step | Action | Math |
| :--- | :--- | :--- |
| **1. Forward** | Compute $Z$ and $A$ for all layers. | $Z = WX+b$, $A=\sigma(Z)$ |
| **2. Loss** | Calculate error. | $L = \text{Loss}(y, \hat{y})$ |
| **3. Backward** | Calculate $\delta$ (local error) from Output to Input. | $\delta_{out} = \frac{\partial L}{\partial A} \sigma'$, $\delta_{hidden} = (W^T \delta_{next}) \sigma'$ |
| **4. Gradients** | Compute weight gradients. | $\frac{\partial L}{\partial W} = \delta \cdot Input^T$ |
| **5. Update** | Adjust weights. | $W \leftarrow W - \alpha \nabla W$ |
