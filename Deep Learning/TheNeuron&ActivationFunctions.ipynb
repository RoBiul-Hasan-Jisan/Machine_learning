# The Neuron & Activation Functions

## 1. The Artificial Neuron (The Perceptron)
An artificial neuron is a mathematical function conceived as a model of biological neurons (brain cells). It is the basic unit of computation in a neural network.

### Anatomy of an Artificial Neuron
The process inside a single neuron happens in two distinct steps:

**1. The Linear Component (Weighted Sum):**
The neuron receives inputs ($x$), multiplies them by specific weights ($w$), and adds a bias ($b$).
* **Weights ($w$):** Determine how important an input is. A higher weight means that input has a greater effect on the output.
* **Bias ($b$):** Allows the neuron to shift the activation function left or right, effectively setting a "threshold" for when the neuron should fire.

Mathematically, this looks like a linear regression equation:
$$z = \sum_{i=1}^{n} (w_i \cdot x_i) + b$$
*Where $z$ is the weighted sum.*

**2. The Non-Linear Component (Activation):**
The result $z$ is passed through an **Activation Function**. This decides whether the neuron should "fire" (pass information to the next layer) and how strong that signal should be.

---

## 2. Why Do We Need Activation Functions?
If we did not use activation functions, the neuron would simply pass on the linear weighted sum ($z$).

**The Critical Rule:** Without non-linear activation functions, a Neural Network is just a Linear Regression model.

No matter how many layers you stack, if they are all linear, the final output is just a linear combination of the inputs. The network would be unable to learn complex patterns like recognizing faces or understanding language.

**Activation functions introduce non-linearity**, allowing the network to learn complex curves and data boundaries.

## 3. Common Activation Functions

### A. Sigmoid (Logistic)
This function takes any real value as input and maps it to a value between **0 and 1**. It looks like an "S" shape.

* **Formula:**
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$
* **Use Case:** Output Layer for binary classification (probability).
* **Cons:** Suffers from the **Vanishing Gradient Problem** (gradients become too small to train the network effectively in deep layers).

### B. Tanh (Hyperbolic Tangent)
Similar to Sigmoid but maps values between **-1 and 1**.

* **Formula:**
$$\tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$$
* **Pros:** It is zero-centered (the average of the data is closer to 0), which makes optimization easier than Sigmoid.
* **Cons:** Still suffers from the Vanishing Gradient Problem.

### C. ReLU (Rectified Linear Unit)
The "King" of modern deep learning. It returns 0 if the input is negative, and returns the input itself if it is positive.

* **Formula:**
$$f(z) = \max(0, z)$$
* **Use Case:** The default choice for **Hidden Layers** (CNNs, MLPs).
* **Pros:** Computationally very fast and solves the Vanishing Gradient issue for positive values.
* **Cons:** **The "Dying ReLU" problem.** If a neuron gets stuck in the negative side, it outputs 0 and stops learning entirely.

### D. Leaky ReLU
A variation of ReLU designed to fix the "Dying ReLU" problem. Instead of being 0 for negative values, it has a small, non-zero slope (e.g., 0.01).

* **Formula:**
$$f(z) = \max(0.01z, z)$$

### E. Softmax
Used exclusively in the **Output Layer** for multi-class classification problems. It converts a vector of raw numbers (logits) into probabilities that sum up to 1.

## Summary Cheat Sheet

| Component | Function |
| :--- | :--- |
| **Input ($x$)** | The raw data fed into the neuron. |
| **Weight ($w$)** | The "knob" the network tunes to learn importance. |
| **Bias ($b$)** | The threshold that shifts the activation. |
| **Sum ($z$)** | The linear combination ($w \cdot x + b$). |
| **Activation ($a$)** | The non-linear decision maker (Sigmoid, ReLU, etc.). |

### Which Activation Function should I use?
1. **Hidden Layers:** Start with **ReLU**.
2. **Output Layer (Binary):** Use **Sigmoid**.
3. **Output Layer (Multi-class):** Use **Softmax**.
4. **Output Layer (Regression):** Use **Linear** (no activation) or ReLU.
