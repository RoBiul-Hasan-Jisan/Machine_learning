# Gaussian Mixture Model 

## 1. What is GMM?
**Gaussian Mixture Model (GMM)** is a probabilistic clustering algorithm based on a core assumption: **Data points are generated from a mixture of several Gaussian (normal) distributions.**

* Each cluster is modeled as its own Gaussian distribution with a specific mean and variance.
* **Key Difference:** Unlike K-Means (which rigidly assigns a point to exactly one cluster), GMM gives a *probability* of belonging to each cluster (known as **soft assignment**). 
* This makes it incredibly useful for datasets where clusters naturally overlap.



---

## 2. Gaussian Distribution Reminder
To understand GMM, you need to understand the underlying math of a 1D Gaussian (Normal) Distribution:

$$f(x) = \frac{1}{\sqrt{2 \pi \sigma^2}} \exp \left( -\frac{(x-\mu)^2}{2\sigma^2} \right)$$

* $\mu$ = mean of the distribution (the center)
* $\sigma^2$ = variance (how wide the spread is)

*Note: For multidimensional data, GMM uses **multivariate Gaussians**, which rely on covariance matrices instead of simple variance.*



---

## 3. How GMM Works (The Intuition)
1. Assume there are $K$ Gaussian distributions hidden in your data.
2. Each cluster has its own mean ($\mu$) and covariance ($\Sigma$).
3. Each point has a probability of belonging to each cluster.

The goal is to maximize the likelihood of the data under these distributions. To find the best parameters, GMM uses the **Expectation-Maximization (EM) algorithm**:
* **E-step (Expectation):** Calculate the probability of each point belonging to each cluster based on current parameters.
* **M-step (Maximization):** Update the Gaussian parameters (mean, covariance, weight) to maximize the overall likelihood.
* **Repeat** these steps until convergence (the parameters stop changing).



---

## 4. Advantages vs. K-Means

| Feature | K-Means | GMM |
| :--- | :--- | :--- |
| **Cluster Assignment** | **Hard** (Point is in exactly 1 cluster) | **Soft** (Probability for each cluster) |
| **Cluster Shape** | Circular / Spherical only | Elliptical / Any Gaussian shape |
| **Overlapping Clusters** |  Fails to separate well |  Handles gracefully |
| **Underlying Math** | Geometric (Distance-based) | Probabilistic (Density-based) |

## 6. Covariance Types 

The covariance_type parameter dictates the geometric shape the clusters are allowed to take:

- **full:** Each cluster has its own covariance matrix (can form any size/angle elliptical clusters).

- **tied:** All clusters share the exact same covariance matrix (same size, shape, and angle).

- **diag:** Only diagonal covariance (ellipses that must align with the X and Y axes; no angled tilting).

- **spherical:** Circular clusters only, very similar to K-Means.


## 7. When to Use GMM
- When your clusters overlap.
- When your clusters are not spherical (e.g., elongated ellipses).
- When you need soft probabilities (e.g., assigning a risk score or mixture modeling).
- For density estimation problems.
**Example Scenario:** 

Customer Segmentation Imagine you are segmenting customers. A customer might buy both gaming gear and cooking supplies.
- **K-Means** would force a 100% assignment to either "Gamer" or "Chef", meaning you lose valuable nuance.
- **GMM** outputs probabilities (e.g., 70% Gamer, 30% Chef). This is vastly more realistic for overlapping real-world groups!

---
## 8. Pros and Cons

**Advantages**

- Handles overlapping clusters effortlessly.
- Provides soft probabilistic clustering (certainty scores).
- Flexible cluster shapes (elliptical).
- Can be used as a generative model for density estimation.

**Disadvantages**

- Needs you to specify the number of clusters $K$ (just like K-Means).
- Highly sensitive to how the starting points are initialized.
- Can be very slow and computationally heavy on massive datasets.
