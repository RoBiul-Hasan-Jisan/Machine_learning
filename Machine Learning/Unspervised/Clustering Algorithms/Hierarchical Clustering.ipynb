# Hierarchical Clustering 

## 1. What is Hierarchical Clustering?
Hierarchical clustering is an unsupervised machine learning algorithm that groups data points into clusters step-by-step, building a tree-like structure. 

Unlike K-Means, you **do not** need to choose the number of clusters ($K$) at the beginning. Instead, the algorithm builds a complete hierarchy of clusters, and you decide later how many clusters you want by analyzing a visual tree diagram called a **Dendrogram**.



---

## 2. Types of Hierarchical Clustering
There are two main approaches to building the hierarchy:

### A. Agglomerative Clustering (Bottom-Up)
This is the most commonly used method in practice.
* **Step 1:** Start with each individual data point acting as its own separate cluster.
* **Step 2:** Find the two closest clusters and merge them into one.
* **Step 3:** Repeat the merging process until only one massive cluster (containing all data points) remains.

*Example Merge Sequence:*
* Step 1: `A`  `B`  `C`  `D`
* Step 2: `(A,B)`  `C`  `D`
* Step 3: `(A,B)`  `(C,D)`
* Step 4: `(A,B,C,D)`

### B. Divisive Clustering (Top-Down)
The exact opposite approach.
* **Step 1:** Start with all data points grouped together in one giant cluster.
* **Step 2:** Split the cluster into two smaller, distinct clusters.
* **Step 3:** Continue recursively splitting until every single data point is its own isolated cluster.



---

## 3. The Dendrogram (Crucial for Evaluation)
A dendrogram is the visual tree diagram that shows exactly how and when clusters merge.

**How to read it:**
* **X-Axis (Bottom):** Represents the individual data points.
* **Y-Axis (Height):** Represents the distance (or dissimilarity) between clusters when they merged. 
* **Lines Joining:** Indicate that a merge has occurred.

**How to choose the number of clusters ($K$):**
To decide on the final clusters, you draw a horizontal "cut" line across the dendrogram. 
* Find the tallest vertical lines that are *not* intersected by any horizontal merge lines.
* Draw your cut line through them. 
* The number of vertical lines your cut intersects equals your final number of clusters.



---

## 4. Linkage Methods (How to define "Closest")
When the algorithm looks for the "two closest clusters" to merge, how does it measure distance between groups of points? This is defined by the **Linkage Criterion**.

* **Single Linkage:** The distance between the *closest* points of two clusters.
  $$Distance = \min(d(a, b))$$ 
  *(Problem: Tends to produce long, drawn-out "chain" clusters).*
* **Complete Linkage:** The distance between the *farthest* points of two clusters.
  $$Distance = \max(d(a, b))$$
  *(Benefit: Produces tight, compact clusters).*
* **Average Linkage:** The average distance between *all pairs* of points in the two clusters.
  *(Benefit: A robust, balanced approach).*
* **Ward Linkage (Very Popular):** Minimizes the variance within the clusters being merged (similar to the K-Means objective function).
  *(Benefit: Usually yields the most highly separated and distinct clusters).*

## 6. Advantages & Disadvantages

### Advantages

- **No initial $K$ required:** You don't have to guess the number of clusters beforehand.
- **Rich Visualization:** The dendrogram provides deep insights into the data's underlying structure.
- **Captures Hierarchy:** Excellent for data that naturally has a taxonomy (e.g., biology, evolutionary trees).
- **Works well for small datasets:** Very accurate and interpretable when $n$ is small.

### Disadvantages
- **Extremely Slow:** Computationally expensive for large datasets.
- **Rigid:** Once a merge is done, it cannot be undone or recalculated later in the process.
- **Sensitive to Noise:** Outliers can severely warp the dendrogram, especially when using Single Linkage.

## 7. Big-O Complexity

Because the algorithm must compute and constantly update a distance matrix between all pairs of points, it is significantly heavier than K-Means.
- **Time Complexity:** $\mathcal{O}(n^2 \log n)$ (in standard implementations)
- **Space Complexity:** $\mathcal{O}(n^2)$ (it must store the $n \times n$ distance matrix)
