# Principal Component Analysis 

## 1. What is PCA?
**Principal Component Analysis (PCA)** is an unsupervised machine learning algorithm used for **Dimensionality Reduction**. 

Instead of just dropping columns, PCA transforms your original features into a brand new, smaller set of features called **Principal Components (PCs)**. 
* These new components capture the maximum amount of "variance" (information) from the original data.
* They are completely uncorrelated (orthogonal) to each other.



---

## 2. When to Use PCA
* **High Dimensionality (The Curse):** When you have hundreds or thousands of features and your model is overfitting or running too slowly.
* **Multicollinearity:** When your features are highly correlated with each other. PCA completely removes this correlation.
* **Data Visualization:** When you have $N$-dimensional data and want to squish it down to 2D or 3D so you can actually plot and look at it.
* **Noise Reduction:** Dropping the lowest-variance components often drops the random noise in your dataset.

---

## 3. How PCA Works (Step-by-Step)
This is a very common interview question. PCA is heavily based on Linear Algebra.

**Step 1: Standardization (Crucial)**
PCA is highly sensitive to the scale of your data. You **must** scale your data so every feature has a mean of 0 and a variance of 1.
$$z = \frac{x - \mu}{\sigma}$$

**Step 2: Compute the Covariance Matrix**
Calculate how each feature varies with every other feature. This creates an $n \times n$ matrix (where $n$ is the number of features).

**Step 3: Calculate Eigenvectors and Eigenvalues**
Perform eigendecomposition on the covariance matrix.
* **Eigenvectors:** The *directions* of the new feature axes (the Principal Components).
* **Eigenvalues:** The *magnitude* (amount of variance/information) carried by each Eigenvector.

**Step 4: Sort and Select**
Sort the Eigenvectors in descending order based on their Eigenvalues. The one with the highest Eigenvalue is **Principal Component 1 (PC1)**. The second highest is **PC2**, and so on. Choose the top $K$ components.

**Step 5: Transform the Data**
Multiply the original standardized data by the chosen Eigenvectors to project the data into the new, lower-dimensional space.



---

## 4. Feature Selection vs. Feature Extraction (PCA)
It is vital to understand the difference:

| Concept | What it does | Example |
| :--- | :--- | :--- |
| **Feature Selection** | Keeps the best original columns, drops the rest. | Keeping `Age` and `Income`, dropping `Height`. |
| **Feature Extraction (PCA)** | Combines original columns into entirely new, uninterpretable variables. | `PC1 = 0.5*Age + 0.3*Income + 0.2*Height` |

## 6. How to Choose 'K' (Explained Variance & Scree Plot)

How do you know how many Principal Components to keep? We look at the Explained Variance Ratio.

- **Explained Variance:** The percentage of the total dataset's information captured by a single Principal Component.

- **Scree Plot:** A line graph plotting the number of components on the X-axis and the cumulative explained variance on the Y-axis.

- **Rule of Thumb:** Find the point on the plot where you retain a high amount of variance (usually 90% to 95%) while significantly reducing the number of dimensions.

## 7. Advantages & Disadvantages

- Advantages

 - Drastically reduces training time for heavy machine learning algorithms.

  -  Solves the multicollinearity problem (PCs are independent).

  - Greatly improves data visualization for high-dimensional data.

  - Filters out background noise and helps prevent overfitting.

- Disadvantages

    - **Loss of Interpretability:** You can no longer explain your model easily. "PC1 increased by 2" means nothing to a business stakeholder compared to "Age increased by 2".

  - **Information Loss:** You are intentionally throwing away some variance (data) to achieve smaller dimensions.

  - **Assumes Linear Relationships:** PCA relies on linear algebra. It struggles to capture complex, non-linear relationships in data (for that, you would use t-SNE or UMAP).
