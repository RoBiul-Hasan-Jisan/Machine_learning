# UMAP (Uniform Manifold Approximation and Projection)

## 1. What is UMAP?
**UMAP** is a state-of-the-art, non-linear dimensionality reduction technique. It was designed to solve the biggest flaws of t-SNE. 

Like t-SNE, it is fantastic for visualizing high-dimensional data in 2D or 3D. However, UMAP is grounded in rigorous mathematical topology, which allows it to preserve **both local structure (who is next to whom) AND global structure (how far apart distinct clusters actually are)**.



---

## 2. Why UMAP over t-SNE? (The Superpowers)
UMAP has largely replaced t-SNE in modern data science workflows for three massive reasons:

1. **Speed:** UMAP is incredibly fast. It can handle massive datasets (millions of rows) in a fraction of the time it takes t-SNE.
2. **Global Structure:** In t-SNE, the distance between two distinct clusters means nothing. In UMAP, if Cluster A is far from Cluster B on the plot, they are actually far apart in the real high-dimensional data.
3. **It Can Transform New Data:** Unlike t-SNE, UMAP has a `transform()` method. You can train a UMAP model on your training data and pass new, unseen test data through it.
4. **Feature Extraction:** Because it preserves global structure and supports any number of output dimensions (not just 2 or 3), you can use UMAP as a preprocessing step *before* passing data to a clustering algorithm (like HDBSCAN or K-Means).

---

## 3. How UMAP Works (The Intuition)
The actual math relies on algebraic topology and Riemannian geometry, but the intuition is straightforward:

* **Step 1: Build a High-Dimensional Graph:** UMAP constructs a "fuzzy" topological representation of your data. It draws edges between points, assuming the data is uniformly distributed on a manifold. Where the data is sparse, the connections stretch further.
* **Step 2: Build a Low-Dimensional Graph:** It creates a similar graph in the lower-dimensional space ( 2D).
* **Step 3: Optimize:** It uses Stochastic Gradient Descent (SGD) to minimize the **Cross-Entropy** between the two graphs, moving the points in the 2D space until the lower-dimensional graph matches the high-dimensional graph as closely as possible.

---

## 4. The Two Most Important Parameters
Tuning UMAP is much more intuitive than tuning t-SNE. You mostly care about two dials:

### A. `n_neighbors` (The Balance Dial)
This dictates how many neighboring points UMAP looks at when learning the manifold structure.
* **Low values ( 5-15):** UMAP focuses intensely on **local structure**. The data will break into many tiny, highly detailed clusters.
* **High values ( 50-200):** UMAP focuses on the **global structure**. You get a broader, more generalized view of the data.

### B. `min_dist` (The Clumpiness Dial)
This controls how tightly UMAP is allowed to pack points together in the final visualization.
* **Low values ( 0.0 - 0.1):** Points will clump tightly together. Great for clustering.
* **High values ( 0.5 - 0.9):** Points will spread out. Great for seeing the internal structure and preventing overlapping blobs.
