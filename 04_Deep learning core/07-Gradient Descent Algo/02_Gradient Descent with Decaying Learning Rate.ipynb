# Gradient Descent with Decaying Learning Rate

Same model as before, but the learning rate is halved whenever an epoch fails to reduce the loss.

```python
import numpy as np
from gd_utils import load_xy, gradient_descent, plot_loss_curve, plot_error_surface
```

## Load data

```python
x, y = load_xy('sgd_data.txt')
print(f'x shape = {x.shape}, y shape = {y.shape}')
```

## Run gradient descent with a decaying learning rate

Each epoch is only accepted if it *improves* the loss; otherwise the learning rate is halved and the epoch is retried. `result.n_iterations` reports how many attempts (successful + retried) that took.

```python
result_mse = gradient_descent(x, y, epochs=500, batch_size=10, loss='mse',
                               variant='decay', lr=0.5, seed=0, verbose_every=100)
print(f'Ran {result_mse.n_iterations} iterations for {len(result_mse.loss_history)} accepted epochs')
```

```python
result_ce = gradient_descent(x, y, epochs=500, batch_size=10, loss='cross_entropy',
                              variant='decay', lr=0.5, seed=0, verbose_every=100)
print(f'Ran {result_ce.n_iterations} iterations for {len(result_ce.loss_history)} accepted epochs')
```

## Loss curves and error surfaces

```python
plot_loss_curve(result_mse, 'Decaying-LR GD -- MSE loss')
plot_loss_curve(result_ce, 'Decaying-LR GD -- cross-entropy loss')
plot_error_surface(x, y, loss='mse', title='MSE error surface')
plot_error_surface(x, y, loss='cross_entropy', title='Cross-entropy error surface')
```
