# Momentum Based Gradient Descent Algorithm

Accumulates a velocity term from past gradients so updates keep moving in a consistent direction. (The original notebook had a bug where the bias update reused the weight's update value -- fixed in `gd_utils.py`.)

```python
import numpy as np
from gd_utils import load_xy, gradient_descent, plot_loss_curve, plot_error_surface
```

## Load data

```python
x, y = load_xy('sgd_data.txt')
print(f'x shape = {x.shape}, y shape = {y.shape}')
```

## Run momentum-based gradient descent

`lr` here plays the role of the momentum decay (often called gamma), and `eta` is the step size applied to the gradient.

```python
result_mse = gradient_descent(x, y, epochs=500, batch_size=10, loss='mse',
                               variant='momentum', lr=0.5, eta=0.1, seed=0, verbose_every=100)
```

```python
result_ce = gradient_descent(x, y, epochs=500, batch_size=10, loss='cross_entropy',
                              variant='momentum', lr=0.5, eta=0.1, seed=0, verbose_every=100)
```

## Loss curves and error surfaces

```python
plot_loss_curve(result_mse, 'Momentum GD -- MSE loss')
plot_loss_curve(result_ce, 'Momentum GD -- cross-entropy loss')
plot_error_surface(x, y, loss='mse', w_range=(-2, 2), b_range=(-2, 2), title='MSE error surface')
plot_error_surface(x, y, loss='cross_entropy', w_range=(-2, 2), b_range=(-2, 2), title='Cross-entropy error surface')
```
